# **VGG-NET Architecture**

- VGG stands for **Visual Geometry Group** models, which is generally used for ***Facial Recognition***, and ***Image Classification***. 
- This model was generally introduced by the Visual Geometry Group from the University of Oxford, 

**Key Components of VGG-19 Architecture:** 
- **Convolutional Layers:** 3x3 filters with a stride of 1 and padding of 1 to preserve spatial resolution.
- **Activation Function:** ***ReLU (Rectified Linear Unit)*** applied after each convolutional layer to introduce non-linearity for improving complex patterns.
- **Pooling Layers:** Max pooling with a 2x2 filter and a stride of 2 to reduce the spatial dimensions.
- **Fully Connected Layers:** Three fully connected layers at the end of the network for classification.
- **Softmax Layer:** Final layer for outputting class probabilities.

**Block 1**
- **Conv1:** 64 filters, 3x3 kernel, ReLU activation
- **Conv2:** 64 filters, 3x3 kernel, ReLU activation
- **Max Pooling:** 2x2 kernel, stride 2

**Block 2**
- **Conv3:** 128 filters, 3x3 kernel, ReLU activation
- **Conv4:** 128 filters, 3x3 kernel, ReLU activation
- **Max Pooling:** 2x2 kernel, stride 2

**Block 3**
- **Conv5:** 256 filters, 3x3 kernel, ReLU activation
- **Conv6:** 256 filters, 3x3 kernel, ReLU activation
- **Conv7:** 256 filters, 3x3 kernel, ReLU activation
- **Conv8:** 256 filters, 3x3 kernel, ReLU activation
- **Max Pooling:** 2x2 kernel, stride 2

**Block 4**
- **Conv9:** 512 filters, 3x3 kernel, ReLU activation
- **Conv10:** 512 filters, 3x3 kernel, ReLU activation
- **Conv11:** 512 filters, 3x3 kernel, ReLU activation
- **Conv12:** 512 filters, 3x3 kernel, ReLU activation
- **Max Pooling:** 2x2 kernel, stride 2

**Block 5**
- **Conv13:** 512 filters, 3x3 kernel, ReLU activation
- **Conv14:** 512 filters, 3x3 kernel, ReLU activation
- **Conv15:** 512 filters, 3x3 kernel, ReLU activation
- **Conv16:** 512 filters, 3x3 kernel, ReLU activation
- **Max Pooling:** 2x2 kernel, stride 2


**Fully Connected Layers**
- **fc6:** 4096 neurons, ReLU activation
- **fc7:** 4096 neurons, ReLU activation
- **fc8:** 1000 neurons, softmax activation (for 1000-class classification)
